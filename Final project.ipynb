# Data visualization for top most popular flight routes 

# Introduction

In the fields of modern aviation, optimizing flight routes stands as a crucial endeavor for airlines seeking efficiency and cost-effectiveness. To address this challenge, this study delves into the analysis and visualization of airline data sourced from the top 10 US airline route segments by volume in 2015. The dataset, obtained from a comprehensive repository of aviation statistics, serves as the foundation for uncovering insights into passenger distribution, route popularity, and geographical considerations that influence route planning decisions.

We use different methods to understand flight routes. Initially, we carefully clean the dataset to make sure it's accurate, including column splitting and removal of redundant information, ensuring the integrity and accuracy of the ensuring analysis. Geographical coordinates for origin and destination cities are then extracted using the Nominatim geocoding service, providing the spatial context necessary for route visualization. The utilization of Matplotlib, Cartopy, Folium, and Bokeh libraries facilitates diverse visualization techniques, ranging from static map plotting to interactive geographical representations and airport location visualizations on a Mercator projection map. These techniques not only offer a comprehensive understanding of flight routes but also enable the identification of key routes and hubs, thus informing strategic decisions regarding route optimization and resource allocation within the airline industry. Through this amalgamation of data analysis and visualization methodologies, this study aims to unravel the intricacies of flight route optimization, ultimately contributing to enhanced operational efficiency and customer satisfaction in the aviation sector.

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
import matplotlib.colors as colors
import holoviews as hv
import pickle
import folium

from bokeh.plotting import figure, show, output_file
from bokeh.models import ColumnDataSource, HoverTool, WMTSTileSource
from bokeh.transform import factor_cmap
from bokeh.palettes import Spectral6
from holoviews import opts
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter
from pyproj import Proj, transform
from matplotlib.cm import ScalarMappable
from math import radians, sin, cos, sqrt, atan2
hv.extension('bokeh')


tile_provider_url = "https://a.basemaps.cartocdn.com/rastertiles/light_all/{Z}/{X}/{Y}.png"
tile_provider = WMTSTileSource(url=tile_provider_url)
ImportWarning



```

```python
file_path = '/Users/berylochieng/Desktop/Projects /Visualization/Top 1000 2015 US Airline Route Segments by Volume.xlsx'
df = pd.read_excel(file_path)
df[['CARRIER_NAME', 'ORIGIN_CITY_NAME', 'DEST_CITY_NAME']]  = df['CARRIER_NAME-ORIGIN_CITY_NAME-DEST_CITY_NAME'].str.split('-', expand=True)

def remove_total_total(city_name):
    if city_name and isinstance(city_name, str):  
        if "Total Total" in city_name:
            city_name = city_name.replace("Total Total", "").strip()
    return city_name


df['DEST_CITY_NAME'] = df['DEST_CITY_NAME'].apply(remove_total_total)
df[['CARRIER_NAME', 'ORIGIN_CITY_NAME', 'DEST_CITY_NAME']]

combined_df = pd.concat([df,df[['CARRIER_NAME', 'ORIGIN_CITY_NAME', 'DEST_CITY_NAME']] ], axis=1)
new_df = combined_df[['PASSENGERS', 'Passenger Miles', 'CARRIER_NAME', 'ORIGIN_CITY_NAME', 'DEST_CITY_NAME']]

geolocator = Nominatim(user_agent="city_locator", timeout = 10)
geocode= RateLimiter(geolocator.geocode, min_delay_seconds=1, max_retries=3, error_wait_seconds=2.0)

cache_path = "city_coordinates.pkl"
if os.path.exists(cache_path):
    with open(cache_path, "rb") as f:
        city_coordinates = pickle.load(f)
else:
    city_coordinates = {}



def get_coordinates(city_name):
    if city_name not in city_coordinates:
        location = geolocator.geocode(city_name)
        if location:
            city_coordinates[city_name] = (location.latitude, location.longitude)
        else:
            city_coordinates[city_name] = (None, None)
    return city_coordinates[city_name]


df['ORIGIN_LATITUDE'] = df['ORIGIN_CITY_NAME'].apply(lambda x: get_coordinates(x)[0])
df['ORIGIN_LONGITUDE'] = df['ORIGIN_CITY_NAME'].apply(lambda x: get_coordinates(x)[1])
df['DEST_LATITUDE'] = df['DEST_CITY_NAME'].apply(lambda x: get_coordinates(x)[0])
df['DEST_LONGITUDE'] = df['DEST_CITY_NAME'].apply(lambda x: get_coordinates(x)[1])

df.to_csv("updated_data.csv", index=False)
file_path = '/Users/berylochieng/Desktop/Projects /Visualization/updated_data.csv'
newdata = pd.read_csv(file_path)
newdata.head()
```

# Geospatial Analysis 
 
 Visualizing the data on map to understand the geographical distribution of passengers traffic and identify popular destinations and understands the regions.

```python
origin_passengers = newdata.groupby('ORIGIN_CITY_NAME').agg({'PASSENGERS': 'sum', 'ORIGIN_LATITUDE': 'mean', 'ORIGIN_LONGITUDE': 'mean'}).reset_index()
m = folium.Map(location=[37.0902, -95.7129], zoom_start=10)
for index, row in origin_passengers.iterrows():
    folium.CircleMarker(
        location=[row['ORIGIN_LATITUDE'], row['ORIGIN_LONGITUDE']],
        radius=row['PASSENGERS'] / 100000,  
        color='blue',
        fill=True,
        fill_color='blue',
        fill_opacity=0.6,
        popup=f"{row['ORIGIN_CITY_NAME']}<br>Total Passengers: {row['PASSENGERS']}"
    ).add_to(m)
m
```

```python
df1 = pd.read_csv("/Users/berylochieng/Desktop/Projects /Visualization/updated_data.csv")  

def haversine(lat1, lon1, lat2, lon2):
    lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
    
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))
    distance = 6371 * c  
    return distance

distances = []
for index, row in df.iterrows():
    origin_lat = row['ORIGIN_LATITUDE']
    origin_lon = row['ORIGIN_LONGITUDE']
    dest_lat = row['DEST_LATITUDE']
    dest_lon = row['DEST_LONGITUDE']
    
    distance = haversine(origin_lat, origin_lon, dest_lat, dest_lon)
    distances.append(distance)

newdata['DISTANCE_KM'] = distances
```

```python
def lon_lat_to_web_mercator(df, lon="LONGITUDE", lat="LATITUDE"):
    merc = Proj(proj="merc", a=6378137, b=6378137)
    x, y = merc(df[lon].values, df[lat].values)
    return x, y


newdata["x_origin"], newdata["y_origin"] = lon_lat_to_web_mercator(newdata, "ORIGIN_LONGITUDE", "ORIGIN_LATITUDE")
newdata["x_dest"], newdata["y_dest"] = lon_lat_to_web_mercator(newdata, "DEST_LONGITUDE", "DEST_LATITUDE")


source_origin = ColumnDataSource(data=dict(
    x=newdata["x_origin"],
    y=newdata["y_origin"],
    city=newdata["ORIGIN_CITY_NAME"],
    passengers=newdata["PASSENGERS"]
))

source_dest = ColumnDataSource(data=dict(
    x=newdata["x_dest"],
    y=newdata["y_dest"],
    city=newdata["DEST_CITY_NAME"],
    passengers=newdata["PASSENGERS"]
))



output_file("airport_plot.html")


p = figure(title="US Airports (Origin & Destination)",
           x_axis_type="mercator", y_axis_type="mercator",
           x_range=(-14000000, -7000000), y_range=(2500000, 7000000),
           width=1000, height=700)

p.add_tile(tile_provider)
p.circle(x="x", y="y", size=6, color="blue", alpha=0.6,
         legend_label="Origin Airports", source=source_origin)
p.circle(x="x", y="y", size=6, color="red", alpha=0.6,
         legend_label="Destination Airports", source=source_dest)


hover = HoverTool(tooltips=[
    ("City", "@city"),
    ("Passengers", "@passengers")
])
p.add_tools(hover)
p.legend.location = "top_left"
p.legend.click_policy = "hide"


show(p)
```

```python
cmap = plt.colormaps.get_cmap("viridis")
norm = colors.Normalize(vmin=newdata["PASSENGERS"].min(), vmax=newdata["PASSENGERS"].max())

fig = plt.figure(figsize=(14, 10))
ax = plt.axes(projection=ccrs.AlbersEqualArea(central_longitude=-96, central_latitude=39))
ax.set_extent([-125, -65, 20, 50], crs=ccrs.Geodetic())


ax.add_feature(cfeature.LAND.with_scale('10m'), facecolor='lightgray')
ax.add_feature(cfeature.STATES.with_scale('10m'), edgecolor='gray', linewidth=0.5)
ax.add_feature(cfeature.BORDERS.with_scale('10m'), linewidth=0.5)
ax.add_feature(cfeature.COASTLINE.with_scale('10m'), linewidth=0.4)


for _, row in newdata.iterrows():
    o_lat, o_lon = row['ORIGIN_LATITUDE'], row['ORIGIN_LONGITUDE']
    d_lat, d_lon = row['DEST_LATITUDE'], row['DEST_LONGITUDE']
    passengers = row['PASSENGERS']
    
    if all(pd.notnull([o_lat, o_lon, d_lat, d_lon])):
        ax.plot([o_lon, d_lon], [o_lat, d_lat],
                color=cmap(norm(passengers)), linewidth=0.6, alpha=0.7,
                transform=ccrs.Geodetic())

hub_data = newdata.groupby("ORIGIN_CITY_NAME")[["PASSENGERS"]].sum().reset_index()
top_hubs = hub_data.sort_values("PASSENGERS", ascending=False).head(10)

for city in top_hubs["ORIGIN_CITY_NAME"]:
    hub_row = newdata[newdata["ORIGIN_CITY_NAME"] == city].iloc[0]
    ax.text(hub_row["ORIGIN_LONGITUDE"], hub_row["ORIGIN_LATITUDE"], city,
            fontsize=8, transform=ccrs.PlateCarree(), color='black', weight='bold')

sm = ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

cbar = fig.colorbar(sm, ax=ax, orientation="horizontal", pad=0.05, aspect=50)
cbar.set_label('Passenger Count (Route Volume)', fontsize=10)

plt.title("U.S. Flight Routes (Color Coded by Passenger Count)\nWith Top 10 Origin Hubs Labeled", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
route_passengers = newdata.groupby(['ORIGIN_CITY_NAME', 'DEST_CITY_NAME']).agg({'PASSENGERS': 'sum'}).reset_index()
route_passenger_miles = newdata.groupby(['ORIGIN_CITY_NAME', 'DEST_CITY_NAME']).agg({'Passenger Miles': 'sum'}).reset_index()

most_popular_routes_passengers = route_passengers.sort_values(by='PASSENGERS', ascending=False).head(10)
most_popular_routes_passenger_miles = route_passenger_miles.sort_values(by='Passenger Miles', ascending=False).head(10)


plt.figure(figsize=(10, 6))
plt.barh(most_popular_routes_passengers['ORIGIN_CITY_NAME'] + ' to ' + most_popular_routes_passengers['DEST_CITY_NAME'], most_popular_routes_passengers['PASSENGERS'])
plt.xlabel('Number of Passengers')
plt.ylabel('Route')
plt.title('Top  Most Popular Routes by Number of Passengers')
plt.gca().invert_yaxis()  
plt.show()

plt.figure(figsize=(10, 6))
plt.barh(most_popular_routes_passenger_miles['ORIGIN_CITY_NAME'] + ' to ' + most_popular_routes_passenger_miles['DEST_CITY_NAME'], most_popular_routes_passenger_miles['Passenger Miles'])
plt.xlabel('Passenger Miles')
plt.ylabel('Route')
plt.title('Top Most Popular Routes by Passenger Miles')
plt.gca().invert_yaxis()  
plt.show()
```
